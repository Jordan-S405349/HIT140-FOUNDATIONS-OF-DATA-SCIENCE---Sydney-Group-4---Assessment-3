# Linear Regression Objective 2.2

### Task

Which team and opponent characteristics explain how many goals this team scores?

### Outcome

`goals_scored` for one team in one match. The file has 208 team–match rows (104 World Cup 2026 matches × 2). Exactly eight predictors, all measured before kick-off, from FIFA, The Stats Don’t Lie, and FBref. Rolling statistics use earlier tournament matches only.

X1–X4 match the ideas in Objective 2.1 (form, attack, opponent defence, host), written for one team. X5–X8 are team draw rate, team failed-to-score rate, opponent loss rate, and opponent clean-sheet rate.

The runnable analysis is `linearreg_objective_2_2.py`. It writes tables and figures to `figures_2_2/`. The report text is `report/Section_2_2_Objective_Report.md`.


### Design choices, aligned with Objective 2.1

- Summary statistics, histograms, boxplots, correlation, VIF, Welch t-test, full OLS, reduced OLS, ridge, lasso.
- Same 5-fold cross-validation: shuffle, `random_state = 42`. Metrics: R², RMSE, MAE.
- Objective 2.1 used ridge α = 1 and lasso α = 0.1. Those penalties are reported again here. A wider penalty search is an extra experiment.
- Objective 2.1 dropped a predictor only when VIF > 5. Here every VIF is below 5. The reduced model instead uses the unit’s |r| ≥ 0.7 screen and drops one variable from each collinear pair.
- A Yeo–Johnson transform of the continuous predictors was tried and did not reduce error.
- GroupKFold by `match_id` checks that pairing the two rows of a match does not change the ranking.
